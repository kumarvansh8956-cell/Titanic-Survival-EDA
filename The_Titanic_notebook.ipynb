# Data cleaning Phase

* detecting the improper data

```python
for c in ds.columns:
  print(c,'\n', ds[c].head())
```

* Dealing with the improperty

```python
ds["fare"] = p.to_numeric(ds["fare"], errors="coerce")
ds["age"] = p.to_numeric(ds["age"], errors="coerce")
```

* Checking for NaN value

```python
print(ds.isnull().sum())
```

* handling the Gender NaN, analysis the Gender data pattern and behaviour

```python
k = ds.groupby("gender")['sn'].count()
print(k)
```

```python
ds["gender"] = ds["gender"].fillna(ds["gender"].mode()[0])
```

-> Expanation :
its means fill the NaN with the value that appears most often.

female    465 [1]
male      835  [0]

Mode means the value that appears most often.

* handling the family NaN, analysis the family data pattern and behaviour

* check for count of families

```python
f = ds.groupby("family")['sn'].count()
print(f)
```

~ explanation
there no any fix pattern, and there only 2 NaN that it effect almost negligible 

```python
ds["family"] = ds["family"].fillna(ds["family"].mode()[0])
```

* handling the pclass NaN, analysis the pclass data pattern and behaviour

```python
p = ds.groupby("pclass")['fare'].count()

print(p)
```

```python
print(ds.loc[ds["fare"].isna()])
```

Still no pattern

```python
ds["fare"]=ds["fare"].fillna(
    ds.groupby("pclass")["fare"].transform("median")
)
```

* Explanation
since their no good patternwe we divide fare in group based on Pclass. and replace the NaN the median of each respective group

* handling the age NaN, analysis the Age data pattern and behaviour

```python
print(ds[ds["age"].isna()])
```

* Expalanation
 the age strogest relatrion this data is title like Mr.,Miss.,Mrs. so first i extracted the title from name

```python
ds["Title"] = ds["Unnamed: 3"].str.extract(r' ([A-Za-z]+)\.')
```

* explanation
Now using this new Title column group up the value of age and replace the Nan with median of respective group

```python
ds["age"] = ds["age"].fillna(
    ds.groupby("Title")["age"].transform("median")
)
```

# Analysis Phase

*Survival Rate

```python
print(" the survival rate; 0 = No , 1 = Yes")

mapping = {0:"death", 1:"survive"}
ds['survived'] = ds['survived'].map(mapping)

survive = ds['survived'].value_counts()
print(survive)
```

* Questions
How many survived?
 500
How many died? 
 800


```python
survival_rate = (500/1300)*100
print(f" survival rate of one person from enitre population on ship is { survival_rate:.2f}%" )
```

```python
mat.pie( survive, labels= survive.index)
mat.title('survival')
mat.show()
```

Reverse mapping

```python
Rev_mapping = {"death":0, "survive":1}
ds['survived'] = ds['survived'].map(Rev_mapping)
```

Reasoning:
we reverse this mapping because this (0,1) is play a important role in our analysis.

# Male and Female survival rate

```python
q = ds.groupby(["gender",'survived'])['sn'].count()

print("the survival rate; 0 = No , 1 = Yes",'\n',q)
```
